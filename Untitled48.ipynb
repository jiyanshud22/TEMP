```python
import os
import cv2
import gc
import torch
from ultralytics import YOLO
from IPython.display import Video, display, HTML
from base64 import b64encode
# import requests # Import requests for more robust downloading - no longer needed for direct download

# Configuration
bytetrack_dir = '/root/Downloads/ByteTrack'
output_videos_dir = os.path.join(bytetrack_dir, 'videos')
os.makedirs(output_videos_dir, exist_ok=True)

# Create pretrained directory
pretrained_dir = os.path.join(bytetrack_dir, 'pretrained')
os.makedirs(pretrained_dir, exist_ok=True)

visdrone_weights_filename = 'yolov8x-visdrone.pt'
visdrone_weights_path = os.path.join(pretrained_dir, visdrone_weights_filename)

# --- IMPORTANT --- #
# The previous download URL for yolov8x-visdrone.pt is no longer accessible (401 Unauthorized error).
# You need to manually download the 'yolov8x-visdrone.pt' file.
# Please upload the downloaded file to the following path in your Colab environment:
# /root/Downloads/ByteTrack/pretrained/yolov8x-visdrone.pt
# --- IMPORTANT --- #

if not os.path.exists(visdrone_weights_path):
    print(f"Error: Model weights file '{visdrone_weights_filename}' not found at '{visdrone_weights_path}'.")
    print("Please manually download the file and place it in the specified directory, then run this cell again.")
    raise FileNotFoundError(f"Model weights file not found: {visdrone_weights_path}")

print(f"Loading model from {visdrone_weights_path}...")
# Initialize Model
visdrone_weights = visdrone_weights_path # Now point to the manually placed path
model = YOLO(visdrone_weights)

# Identify ALL Files in /content/
video_extensions = ['.mp4', '.avi', '.mov', '.mkv']
video_files = []
for f in os.listdir('/content/'):
    if os.path.splitext(f)[1].lower() in video_extensions:
        video_files.append(os.path.join('/content/', f))

# Resume Logic
if 'successfully_processed_videos' not in locals(): successfully_processed_videos = []
if 'skipped_videos' not in locals(): skipped_videos = []

print(f'Total videos identified in /content/: {len(video_files)}')

for idx, video_path in enumerate(video_files):
    fname = os.path.basename(video_path)
    h264_path = os.path.join(output_videos_dir, f'{os.path.splitext(fname)[0]}_tracked_h264.mp4')

    if os.path.exists(h264_path):
        print(f'>>> {fname} already exists. Skipping.')
        if video_path not in successfully_processed_videos: successfully_processed_videos.append(video_path)
        continue

    print(f'\n--- STARTING VIDEO {idx+1}/{len(video_files)}: {fname} ---')
    try:
        cap = cv2.VideoCapture(video_path)
        fps = int(cap.get(cv2.CAP_PROP_FPS))
        w = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
        h = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
        total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
        cap.release()

        raw_out = h264_path.replace('_h264.mp4', '.mp4')
        writer = cv2.VideoWriter(raw_out, cv2.VideoWriter_fourcc(*'mp4v'), fps, (w, h))

        # Tracking loop with persist=True to maintain IDs
        results = model.track(source=video_path, tracker='bytetrack.yaml', device='cpu', imgsz=640, conf=0.2, persist=True, stream=True, verbose=False)

        frame_idx = 0
        for r in results:
            frame = r.orig_img.copy()
            if r.boxes is not None and r.boxes.id is not None:
                # Get boxes, classes, and track IDs
                boxes = r.boxes.xyxy.int().cpu().tolist()
                cls = r.boxes.cls.int().cpu().tolist()
                ids = r.boxes.id.int().cpu().tolist()

                for box, class_id, track_id in zip(boxes, cls, ids):
                    x1, y1, x2, y2 = box
                    label = f'{model.names[class_id]} ID:{track_id}'

                    # Draw bounding box
                    cv2.rectangle(frame, (x1, y1), (x2, y2), (0, 255, 0), 2)
                    # Draw label
                    cv2.putText(frame, label, (x1, y1 - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)

            writer.write(frame)
            frame_idx += 1
            if frame_idx % 10 == 0:
                print(f'  -> {fname}: Frame {frame_idx}/{total_frames} processed...', end='\r')

        print(f'\n  -> {fname}: Frame processing complete.')
        writer.release()

        print(f'  -> {fname}: Converting to H.264...')
        !ffmpeg -y -i "{raw_out}" -vcodec libx264 -acodec aac "{h264_path}" -loglevel quiet

        if os.path.exists(raw_out): os.remove(raw_out)
        successfully_processed_videos.append(video_path)
        print(f'✅ SUCCESS: {fname} finished.')

    except Exception as e:
        print(f'\n❌ ERROR on {fname}: {e}')
        skipped_videos.append(video_path)

    gc.collect()
    torch.cuda.empty_cache() if torch.cuda.is_available() else None

print('\n--- FINAL BATCH STATUS ---')
print(f'Processed: {len(successfully_processed_videos)} | Skipped: {len(skipped_videos)}')
```
